# Trimming and decontamination from Human Genomic using Kneaddata
To get the microbiome only sequence, the human sequence must be decontaminated using Kneaddata (https://github.com/biobakery/kneaddata) and Bowtie2 using GRCh38.p14 as host reference and trimming using Trimmomatic (https://github.com/usadellab/Trimmomatic) with parameter adapter-trimmed and quality-filtered (Phred <20, minimum length 50 bp), Trimmomatic version 0.39, additional parameters are SLIDINGWINDOW:4:20 MINLEN:50 LEADING:3 TRAILING:3. 


```python
kneaddata --input1 File_1.fastq.gz --input2 File_2.fastq.gz -db </path/to/human_ref> --output File_kneaddata_output --trimmomatic </path/to/Trimmomatic> --trimmomatic-options= SLIDINGWINDOW:4:20 MINLEN:50 --sequencer-source TruSeq3 --threads 10 
```

# Taxonomic assignment of metagenomics sequencing reads using Kraken2
Kraken2 installation and source code (https://github.com/DerrickWood/kraken2) and Kraken2 databases can be downloaded from https://benlangmead.github.io/aws-indexes/k2

```python
k2 classify --db archaea,viral,univec --output output.txt --threads 9 --unclassified-out unclassified.fasta --classified-out 'classified.fasta' --report report.txt reads.fna 
```

# Microbiome profiling and annotation using Metaphlan4
Metaphlan4 installation and source code (https://github.com/biobakery/biobakery/wiki/metaphlan4) 


```python
metaphlan file.fasta.gz --input_type fasta > file_profile.txt 
```
